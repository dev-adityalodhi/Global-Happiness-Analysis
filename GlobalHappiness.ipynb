# Global Happiness Data Analysis
* This program analyzes global happiness data from a CSV file and provides insights such as average happiness scores, regional comparisons, and trends over time.

# 🧠 Concept:
To explore the World Happiness Report dataset, which contains metrics for around 158 countries:

* Country

* Happiness Score

* GDP per Capita

* Social Support

* Healthy Life Expectancy

* Freedom

* Generosity

* Perceptions of Corruption

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('D:/python/data/happiness.csv')
print(type(df))
df
```

```python
df.head()
```

```python
df.dropna()
```

```python
df.sort_values('Economy (GDP per Capita)', ascending=False)
```

```python
mean = df['Happiness Rank'].mean()
mean
```

```python
sd = df['Happiness Rank'].std()
sd
```

```python
happiness = df['Happiness Rank']
gdp = df['Economy (GDP per Capita)']
score = df['Happiness Score']
corr = np.corrcoef(happiness, gdp)[0, 1]
corr2 = np.corrcoef(happiness, score)[0, 1]
corr, corr2
```

```python
top_10 = df.nsmallest(10, 'Happiness Rank') # 10 lowest ranks
top_10
```

```python
bottom_10 = df.nlargest(10, 'Happiness Rank') # 10 highest ranks
bottom_10
```

```python
top_happiness_score = df.nlargest(10, 'Happiness Score') # 10 highest scores
top_happiness_score
```

```python
bottom_happiness_score = df.nsmallest(10, 'Happiness Score') # 10 lowest scores
bottom_happiness_score
```

```python
regional_avg = (
    df.groupby('Region')['Happiness Score']
      .mean()
      .reset_index()   # ✅ converts from Series → DataFrame
      .sort_values(by='Happiness Score', ascending=False) # ✅ sort descending
)
regional_avg
```

```python
import os
os.makedirs('charts', exist_ok=True)

plt.figure(figsize=(10,6))
plt.barh(regional_avg["Region"], regional_avg["Happiness Score"], color='skyblue')
plt.xlabel("Average Happiness Score")
plt.ylabel("Region")
plt.title("Average Happiness Score by Region")
plt.gca().invert_yaxis()  # happiest at the top
plt.savefig('charts/happiness_by_region.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df["Economy (GDP per Capita)"], df["Happiness Score"], color='red', alpha=0.6) # scatter plot
plt.title("Happiness Score vs Economy (GDP per Capita)")
plt.xlabel("Economy (GDP per Capital")
plt.ylabel("Happiness Score")
plt.grid(True)
plt.savefig('charts/happiness_vs_gdp.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.bar(top_10["Country"], top_10['Happiness Rank'], color = 'orange', alpha = 1) # alpha for transparency
plt.xlabel("Country")
plt.ylabel("Happiness Rank")
plt.title("Top 10 Happiest Countries")
plt.savefig('charts/happiness_bar.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
happiness_score = df['Happiness Score']
plt.figure(figsize=(12,6))
plt.boxplot(happiness_score)
plt.title("Distribution of Happiness Scores")
plt.ylabel("Happiness Score")
plt.savefig('charts/happiness_boxplot.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(regional_avg['Region'], regional_avg['Happiness Score'],
         marker='D', linestyle='--', color='orange', linewidth=2)
plt.xticks(rotation=90, ha='right')
plt.xlabel('Region')
plt.ylabel('Average Happiness Score')
plt.title('Average Happiness Score by Region')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('charts/happiness_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
